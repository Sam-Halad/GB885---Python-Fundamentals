A competitive swimwear company is launching a swimwear line for taller women. The compnay wants to limit the search to 2016 competitors. Identify the talles medalist from the 2016 games.



```python
#find tallest medalist from 2016 games
#filter for 2016
year_filter = female_swimmers[female_swimmers['Year'] == 2016]
#drop rows where "Medal" is NaN
female_swimmer_medalist = year_filter.dropna(subset=['Medal'])
#find the tallest swimmer
tallest_swimmer = female_swimmer_medalist.sort_values(by='Height', ascending=False).iloc[0]
print(tallest_swimmer)
```

The strategy team is interested in trends among highest paid athletes. The following GitHub repository has a dataset of 100 highest paid athletes. Load the Athletes.csv files from GitHub and answer the following questions

```python
#load the GitHub repository
url = 'https://raw.githubusercontent.com/ali-ce/datasets/refs/heads/master/Most-paid-athletes/Athletes.csv'
athlete_salaries = pd.read_csv(url)
athlete_salaries.head()
```

```python
athlete_salaries['Total Pay'] = athlete_salaries['Total Pay'].str.replace('$', '')

athlete_salaries['Total Pay'] = athlete_salaries['Total Pay'].str.replace(',', '')
#convert to numeric
athlete_salaries['Total Pay'] = pd.to_numeric(athlete_salaries['Total Pay'])
athlete_salaries.head()
```

```python
# what is the correlation between an athlete's height and total pay
correlation = athlete_salaries['Height (cm)'].corr(athlete_salaries['Total Pay'])
print(correlation)
```

```python
#what is the difference between 75th percentile and 25th percentile of top-paid athlete ages
#calculate age. We only have date of birth so take a reference year and subtract last 4 digits of birth date
ref_year = 2025
athlete_salaries['Age'] = ref_year - athlete_salaries['Birth Date'].str[-4:].astype(int)
#calculate quanitles
iq3 = athlete_salaries['Age'].quantile(.75)
iq1 = athlete_salaries['Age'].quantile(.25)
#calculate iqr
iqr = (iq3-iq1)
print(f"The Interquartile range of ages is: {iqr}")
```

```python
#compare the mean total pay by sport
sport = athlete_salaries.groupby('Sport')
#calculate mean and sort
mean_salary = sport['Total Pay'].mean()
salary_sort = mean_salary.sort_values(ascending=False)
#print and divide to convert to millions
print(f"The mean salary by sport in millions ($) {salary_sort/1000000}")
```

The basketball team has an API that will return statistics by season. Use the API to call data from the 2024 season to help the team with contract negotiations.

```python
import requests
import json
basketball_url = 'https://api.server.nbaapi.com/api/playertotals?season=2024&page=1&pageSize=1000'
#get data from the api
response = requests.get(basketball_url)
bball_data = response.json()
bball_df = pd.DataFrame(bball_data['data'])

bball_df.head()
```

```python
#what is the maximum number of games started by a pleaser in 2024
games_started = bball_df.sort_values(by = 'gamesStarted', ascending = False)
print(games_started.iloc[0]['gamesStarted'])
#how many players started that number of games?
max_games_count = games_started.iloc[0].count()
print(f"{max_games_count} players started in the maximum number of games in the 2024 season")
```

Create two new features: Stocks per game that calculates each player's stocks divided by games played in the 2024 season. "minutes_per_game" that calculates each player's average minutes played in a game. Create a list of 10 underused players with the highest "Stocks per Game".

```python
#create stocks per game column
bball_df['stocks_per_game'] = (bball_df['steals'] + bball_df['blocks']) / bball_df['games']
#create minute per game column
bball_df['minutes_per_game'] = bball_df['minutesPg'] / bball_df['games']
bball_df.head()
```

```python
#create a list of 10 underused players with highest stocks per game.
#filter underused players as less than 25 minutes per game
underused_df=bball_df[bball_df['minutes_per_game'] <25]
stocks_sorted = underused_df.sort_values(by='stocks_per_game', ascending=False)
print(stocks_sorted.head(10))
#locate the player with highest stocks per game
stocks_sorted.iloc[0]['playerName']
```
