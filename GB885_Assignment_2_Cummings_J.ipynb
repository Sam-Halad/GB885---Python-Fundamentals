```python
#create a for loop that calculates bonus for all bank tellers
employees = [
    {'name': 'Alice', 'rating':4.88, 'salary': 67000},
    {'name': 'Bob', 'rating':4.2, 'salary': 52300},
    {'name': 'Charlie', 'rating':3.26, 'salary': 46000},
    {'name': 'David', 'rating':2.33, 'salary': 45200},
    {'name': 'Eve', 'rating':1.06, 'salary': 35100}]
#for index in employees
for i in employees:
  #loop over each index to find rating and salary of each employee
    rating = i['rating']
    salary = i['salary']

  #use if statement from previously created conditional statment for each rating and salary
    if rating >= 4.5 and rating <= 5:
        bonus = salary * 0.17
    elif rating >= 4 and rating < 4.5:
        bonus = salary * 0.1
    elif rating >= 3 and rating < 4:
        bonus = salary * 0.05
    elif rating >= 1.5 and rating < 3:
        bonus = salary * 0.02
    elif rating >= 0 and rating < 1.5:
        bonus = 0

    print(f"{i['name']} has a bonus of ${bonus}")
```
