You are an analyst for a global infant and maternal health advocacy group called JPG. Josephina, the founder, is preparing a presentation for the UN and has asked you to pull some insights for her from the data.

```python
import pandas as pd
import numpy as np
#create pandas dataframe through url provided
#Read the CSV file into a pandas dataframe.
url = 'https://raw.githubusercontent.com/katiegaertner/ML-Fundamentals/main/world-health-data-numbers.csv'
df_health = pd.read_csv(url)
```

```python
#how many rows are in the DataFrame
df_health.shape
#df_health.head()
```

View the first five rows of the DataFrame. There is a column called 'Abbreviations', which holds a unique two-character value for each country. Set 'Abbreviations' as the index. Sort the dataframe by this new index.

```python
#view first 5 entries
df_health.head(5)
#set abbreviaiton as index
df_health = df_health.set_index('Abbreviation')
#sort index
df_health = df_health.sort_index()
```

```python
#What country uses 'DZ' as abbreviation
df_health.loc['DZ']
```

We do not want to include Vatican City in our DataFrame due to their birthing process. Delete Vatican city from the dataframe.

```python
#delete Vatican City from the data
df_health.drop('VA')
```

```python
# what is the official language of the nation of the 186th row. Use 185 since Python starts at 0
df_health.iloc[185]
print(df_health['Official Language'].iloc[185])
```

```python
#sort the dataframe by infant mortality rate.
df_health_copy= df_health.sort_values('Infant mortality_per_1k', ascending = False)
df_health_copy.head(10)
```

```python
#create a new dataframe that includes all the columns, but only for countries whose offical language is Spanish
df_health_spanish= df_health[df_health['Official Language'] =='Spanish']
df_health_spanish.head()
df_health_spanish.shape
```

```python
#what is the mean infant mortality per 1000 infants in countries whose offical language is spanish
round(df_health_spanish['Infant mortality_per_1k'].mean(),2)
```

```python
#what is the median maternal mortality in countries whose official language is spanish?
df_health_spanish['Maternal_Mortality_per_100k'].median()
```

```python
#create a new column for GDP per capita and find the nation with fourth highest GDP per capita. Divide GDP by Population size
df_health['GDP_per_capita'] = df_health['GDP']/df_health['Population']
#sort by gdp per capita
df_health_gdp_sort = df_health.sort_values('GDP_per_capita', ascending=False)
#df_health_gdp_sort.head()
#find the fourth country using index 3 because python starts at 0
fourth_country = df_health_gdp_sort.iloc[3]
print(fourth_country)
#print that country's maternal mortality rate
print(fourth_country['Maternal_Mortality_per_100k'])
```

```python
#what is the average maternal mortality rate per 100k for all nations
round(df_health['Maternal_Mortality_per_100k'].mean(),2)
```

```python
#what percent of health care is paid out of pocket by country with the lowest infant mortaltiy
lowest_infant_mortality = df_health.sort_values('Infant mortality_per_1k').iloc[0]
print(lowest_infant_mortality)
print(lowest_infant_mortality['oop_health_exp'])
```

```python
#what perent of health care is paid out of pocket by the country with the highest infant mortality?
highest_infant_mortality = df_health.sort_values('Infant mortality_per_1k',ascending=False).iloc[0]
print(highest_infant_mortality)
print(highest_infant_mortality['oop_health_exp'])
```
