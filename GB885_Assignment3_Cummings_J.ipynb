We need to develop a function that allows customers to place an order at a kiosk, and adjust our inventory.

```python
monday_menu = {'choccy chipper':{'price':2.50,'stock':120},
               'fudgy pudgy':{'price':2.50,'stock':120},
               'key lime crime':{'price':4.25,'stock':96},
               'salty sailor':{'price':4.40,'stock':60},
              'blueberry blooper':{'price':5.0,'stock':60}}

def cookie_inventory(menu):
  item = str(input('Input the cookie: '))
  quantity = int(input('input the number of cookies: '))

#write a while loop to keep asking the user to input a cookie until they choose one on the menu
  while item not in menu:
    print(f'The cookie is not on the menu. Please order a cookie that is on the menu')
    item = str(input('Input the cookie: '))
    quantity = int(input('input the number of cookies: '))
  stock = menu[item]['stock']
  #check the inventory
  if stock - quantity < 0:
    print(f'The cookie is out of stock. We only have {stock} of this type left')


  #write an input statement that asks customers if they would like to order a different amount.

    change_quantity = input("Would you like to order a different amount? (Y/N): ")

    if change_quantity == 'Y':
      quantity = int(input(f'Enter a new quantity for {item}'))
      if menu[item]['stock'] - quantity < 0:
        print(f'You still ordered too many. We only have {quantity} Please enter a number less than or equal to the number in stock')
      else:
      #subtract quantity ordered from menu
        menu[item]['stock']=menu[item]['stock']-quantity
      #calculate order total
        total = menu[item]['price'] * quantity
        print(f'You ordered {quantity} {item}s for ${total:.2f} Thank you for your order')
  #ask the user to input whether they would like to order a differetn cookie
    else:
      change_cookie = input('Would you like to order a different cookie? (Y/N): ')
      if change_cookie == 'Y':
          print('Please restart your order')
      else:
        print('Thank your for your order. Try again')
  else:
      menu[item]['stock']=menu[item]['stock'] - quantity
      total = menu[item]['price'] * quantity
      print(f' You ordered {quantity} {item}s for ${total:.2f}. Thank you')

cookie_inventory(monday_menu)

```
