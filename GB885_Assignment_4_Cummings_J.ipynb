We are a payroll anaylst for a tutoring service called SmartyPants. The tutors work at an hourly rate. You have inherited code from the previous analyst, who was attempting to create a program that allows tutors to enter their shift hours, allows payroll to calculate the pay period salary and allows payroll to display all tutors and their pay period salary at once. We are tasked with debugging the code and devloping a payroll system that meets the project objectives.

```python
#inherited code
employees = {'Jan': {'hourly_rate': 23.50, 'hours': 0},
            'Maria': {'hourly_rate': 35.25, 'hours': 0},#missing curly brace
            'Renee': {'hourly_rate': 30.50, 'hours': 0},
            'Haruka': {'hourly_rate': 38.00, 'hours': 0},#found missing comma
            'Samir': {'hourly_rate': 24.75, 'hours': 0}}
```

```python
#paste the code and rename the function.
def weekly_hours_update(name, shift_hours):
    new_hours = employees[name]['hours'] + shift_hours
    employees[name]['hours'] = new_hours
```

```python
# test the function
weekly_hours_update('Renee',8)
print(f" Renee's timecard has updated. They have worked: {employees['Renee']['hours']} hours this week")
```

```python
#Call your function six times with parameters as shown.
weekly_hours_update('Samir',5)
weekly_hours_update('Maria',4)
weekly_hours_update('Jan',10)
weekly_hours_update('Haruka',4)
weekly_hours_update('Maria',3)
weekly_hours_update('Haruka',2)
print(employees)
```

```python
def calculate_salary(name):
  return employees[name]['hourly_rate'] * employees[name]['hours']
```

```python
#call the function with parameters
print(f"Jan's total salary for this week is: ${calculate_salary('Jan')}")
```

```python
#pasted function
def total_salary():
  for employee in employees:
    salary = calculate_salary(employee)#function name has to match previously defined function
    print(f'{employee} has earned ${salary} this pay period')#the variable 'salary' needs to be in curly brackets

total_salary()
```
